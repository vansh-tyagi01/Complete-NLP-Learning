# Step 2 [Removing Unimportant info like HTML tag]

```python
# Remove HTML tags in text string.

import re
def remove_html_tags(text):
    pattern = re.compile('<.*?>')
    return pattern.sub(r'',text)

text = "<html><body><p>Movie 1</p><p> Actor - Aamir khan </p><p>Click here to <a href='http://google.com'>download</a></p></p>"
```

```python
remove_html_tags(text)
```

```python
# Apply in review column.

df['review'] = df['review'].apply(remove_html_tags)
```

# Step 3 [remove URLs]

```python
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'',text)
```

```python
text_url = "Visit our website at https://www.example.com to learn more about NLP."
```

```python
remove_url(text_url)
```

# Step 4 [Remove punctuation]

```python
import string,time
string.punctuation
```

```python
exclude = string.punctuation
```

```python
# Slow method

def remove_punc(text):
    for char in exclude:
        text=text.replace(char,'')
    return text
```

```python
text_punc = "string. With. punctuation?"
```

```python
start = time.time()
print(remove_punc(text_punc))
time1=time.time() - start
print(time1*50000)  # Means remove total no of raws punctuation take a time 29 sec.
```

```python
# Faster method

def remove_punc_fast(text):
    return text.translate(str.maketrans('', '', exclude))
```

```python
start = time.time()
print(remove_punc_fast(text_punc))
time2=time.time() - start
print(time2*50000) 
```

# Step 5 [chat word treatment]

```python
chat_words = {
    'FYI':'For your information',
    'GN':'Good night'
}
```

```python
def chat_conversion(text):
    new_text = []
    for w in text.split():
        if w.upper() in chat_words:
            new_text.append(chat_words[w.upper()])
        else:
            new_text.append(w)
    return " ".join(new_text)
```

```python
chat_conversion('FYI thanks')
chat_conversion('GN babu')
```

# Step 7 [Removing stop words]

```python
from nltk.corpus import stopwords
```

```python
stopwords.words('english')
```

```python
def remove_stopwords(text):
    new_text=[]
    for word in text.split():
        if word in stopwords.words('english'):
            new_text.append('')
        else:
            new_text.append(word)
    x = new_text[:]     # copy bnata he jo new text me ho vhi.
    new_text.clear()    # new text ko khali kar diya.
    return " ".join(x)   # Ab x me he sara jo new text me tha to x se ek sentance bna diya sare split word ka.
    
```

```python
remove_stopwords('This is a dog')
```

```python
# If apply in our dataset.

df['review'] = df['review'].apply(remove_stopwords)
```
